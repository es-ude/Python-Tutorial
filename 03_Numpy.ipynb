## Arrays anlegen

Ein Wichtige Aspekt besteht in der Interpretation von Daten als Matrize (oder Array).
Diese Arrays können mit NumPy einfach angelegt werden.

Es handelt sich hierbei um eine spezielle Form von Listen.
Eine wichtige Eigenschaft dieser NumPy Arrays besteht darin, dass sie nur Element **eines** fest definierten Datentyps enthalten können.
Die zu Verfügung stehenden Datentypen werden hierbei durch NumPy vorgegeben

- Vorzeichen behaftete Integer Datentypen:
    - `int8` -> signed 8 Bit Ganzzahl (char)
    - `int16` -> signed 16 Bit Ganzzahl (short)
    - `int32` -> signed 32 Bit Ganzzahl (int)
    - `int64` -> signed 64 Bit Ganzzahl (long)
- Vorzeichen freie Integer Datentypen:
    - `bool` -> Boolean
    - `int8` -> unsigned 8 Bit Ganzzahl (uchar)
    - `int16` -> unsigned 16 Bit Ganzzahl (ushort)
    - `int32` -> unsigned 32 Bit Ganzzahl (uint)
    - `int64` -> unsigned 64 Bit Ganzzahl (ulong)
- Fließkommazahlen:
    - `float16` -> Half precision Float (16 Bit)
    - `float32` -> Float (32 Bit)
    - `float64` -> Double (64 Bit)
    - `float128` -> Long Double (128 Bit)
- Komplexe Zahlen:
    - `complex64` -> Komplexe Zahl (Dargestellt durch zwei 32 Bit Floats)
    - `complex128` -> Komplexe Zahl (Dargestellt durch zwei 64 Bit Double)
    - `complex256` -> Komplexe Zahl (Dargestellt durch zwei 128 Bit Long Double)

Der Datentyp eines Arrays lässt sich mit Hilfe der Funktion

```python
numpy_array.astype(np.numpy_type)
```

konvertieren.

Um ein Array anzulegen stehen mehrere Funktionen zur Verfügung

- `np.zeros((x,y))` -> Array mit x Zeilen und y Spalten, wobei alle Elemente 0 sind
- `np.ones((x,y))` -> Array mit x Zeilen und y Spalten, wobei alle Elemente 1 sind
- `np.eye(w)` -> (Quadratische) 2D Identitätsmatrix 
- `np.arrange(start, end, step)` -> Array mit Werten von _start_ bis _end_ in Schritten ger Größe _step_
- `np.asarray([...])` -> Konvertiert eine Liste von Zahlen in ein NumPy Array
- ...

## Daten manipulieren

NumPy bietet viele Funktionen um Daten zu manipulieren.

- Addition: `numpy_array_1 + numpy_array_2`
- Multiplikation: `np.dot(numpy_array_1, numpy_array_2)`
- Transposition: `np.transpose(numpy_array_1)`
- Reshape: `np.reshape(numpy_array_1, (axis_1, ...))`
- Diagonale ersetzen: `np.diag()`

> **TIP**:
> Um einen Zeilenvektor in einen Spaltenvektor umzuwandeln muss das Array **zwei** dimensional sein!
> Mit der Funktion `np.atleast_2d()` kann dies sichergestellt werden.
