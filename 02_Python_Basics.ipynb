### Variablen

Variablen haben in Python keine feste Typenzuweisung.
Sie werden dynamisch bei der Zuweisung von Werten festgelegt.

Sollte einer Variablen ein Wert von einem anderen Typ zugewiesen werden ist dies ohne Probleme möglich, da der Interpreter dies nicht kontrolliert.
Um eine gewisse Typsicherheit zu gewährleisten werden daher häufig spezielle Linter wie [mypy](https://www.mypy-lang.org) genutzt, welche den Typ einer Variable mit den zugewiesenen Werten vergleichen. 
Damit Typechecker wie mypy wissen welchen Typ eine Variable haben sollte kann man sogenannte _type hints_ an den Namen einer Variable schreiben.

```python
var_name: type-hint = value
```

Für das anlegen einer Variable ist das Zuweisen eines Wertes optional!

**Einfache _type hints_:**

|   Typ | Erklärung                                              |
| ----: | :----------------------------------------------------- |
|  None | Spezielle Wert, der angibt das es keinen Wert gibt.    |
|   int | Ganzzahlige Werte (Integer, Long, ...)                 |
| float | Fließkommazahlen (Float, Double, ...)                  |
|  bool | Wahrheitswerte (Boolean: True, False)                  |
|   str | Zeichenketten (Character, String, ...)                 |
| bytes | Bytefolge (ASCII wird automatisch in Byte konvertiert) |

**Komplexe _type hints_:**
|   Typ | Erklärung |
| ----: | :- |
|  dict | Dictionary. Der Typ von Schlüssel und Wert kann in eckigen Klammern angegeben werden: `dict[str,int]` -> Schlüssel: str, Werte: int    |
|  list | Ungeordnete Liste. Der Typ der Elemente kann in Eckigen Klammern angegeben werden: `list[int,float]` -> Liste mit int und float Werten |
|   set | Geordnete Liste mit variabler Länge. Der Typ der Elemente kann in Eckigen Klammern angegeben werden: `set[int,float]` -> Set mit int und float Werten |
| tuple | Geordnete Liste mit fester Länge. Der Typ der Elemente kann in Eckigen Klammern angegeben werden: `set[int, ...]` -> Set mit int Werten, `set[float, str]` -> Set mit float und str Werte |

### Listen

Eigenschaften aller Listen in Python:
- Die Indexierung startet bei 0.
- Elemente können einzeln oder in slices (Bereichen) abgerufen werden.
- Der Datentyp innerhalb einer Liste kann variieren

Es gibt drei wesentliche Arten von Listen:
- Ungeordnete Listen variabler Länge: List
- Geordnete Listen variabler Länge: Set
- Geordnete Listen fester Länge: Tupel

#### List

```python
list_name: list[type_hints] = [wert1, ...]
```
- Werden keine Werte angegeben handelt es sich um eine leere Liste.
- Es können mehrere type hints angegeben werden.

**Wichtige Funktionen:**
- Anfügen eines Eintrags: `list_name.append(wert)`
- Anfügen mehrerer Einträge: `list_name += [wert1, wert2, ...]`
- Länge der Liste: `len(list_name)`
- Abrufen eines Elementes: `list_name[index]`
- Abrufen eines Bereiches: `list_name[index_start:index_end]` (exklusive `index_end`)

````python
# Legen sie eine Liste mit den Elementen 'Hello', 'World' an

hello: ...
````

#### Tupel

```python
tuple_name: tuple[type_hints] = (wert1, ...)
```
- Werden keine Werte angegeben handelt es sich um ein leeres Tuple.
- Es können type hints für jedes Element angegeben werden
    - Das Tuple hat dann eine feste Länge
    - Der type hint gibt den Type des Wertes mit dem selben index an
- Es kann ein type hint gefolgt von `, ...` angegeben werden
    - Die Länge des Tupel ist variable
    - Jeder Wert hat den Typ des type hints

**Wichtige Funktionen:**
- Länge der Liste: `len(tuple_name)`
- Abrufen eines Elementes: `tuple_name[index]`
- Abrufen eines Bereiches: `tuple_name[index_start:index_end]` (exklusive `index_end`)

### Dictionaries

```python
dict_name: [type_hint_key, type_hints_values] = {schlüssel1: wert1, schlüssel2: ...}
```

- Werden keine Werte angegeben, so ist das Dictionary leer
- Der Typ des Schlüssels ist im gesamten Dictionary gleich
- Der Typ des Wertes kann variabel sein
    - Die type hints werden in der Form `type_hint | type_hint | ...` angegeben

**Wichtige Funktionen:**
- Einen Eintrag einfügen/ändern: `dict_name[schlüssel] = wert`
- Mehrere Einträge einfügen/ändern: `dict_name.update(dict_to_update)`
- Länge des Dictionaries: `len(dict_name)`
- Zugriff auf einen Eintrag: `dict_name[schlüssel]`
- Liste aller schlüssel: `dict_name.keys()`
- Liste aller Werte: `dict_name.values()`
- Liste aller Einträge als Tupel: `dict_name.items()`
